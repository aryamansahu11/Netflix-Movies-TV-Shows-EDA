```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/netflix_titles.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
# 1. Dataset shape
print("Dataset Shape:", df.shape)

# 2. First 5 rows
display(df.head())

# 3. Last 5 rows
display(df.tail())

# 4. Column information
df.info()

# 5. Data types
print(df.dtypes)

# 6. Memory usage
print("Memory Usage:")
print(df.memory_usage(deep=True))
```

```python
# Count missing values
missing = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (missing / len(df)) * 100

# Create missing values table
missing_df = pd.DataFrame({
    "Missing Count": missing,
    "Missing Percentage": missing_percentage
})

display(missing_df.sort_values(
    by="Missing Count",
    ascending=False
))
```

```python
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(
    ascending=False
)

plt.figure(figsize=(10, 5))
plt.bar(missing.index, missing.values)

plt.title("Missing Values by Column")
plt.xlabel("Column Names")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Summary statistics of release year
print("Release Year Statistics")
display(df["release_year"].describe())

# Mean, median and standard deviation
print("Mean:", df["release_year"].mean())
print("Median:", df["release_year"].median())
print("Standard Deviation:", df["release_year"].std())

# Quartiles
print("Quartiles:")
print(df["release_year"].quantile([0.25, 0.50, 0.75]))
```

```python
# Content type
print("Content Type:")
display(df["type"].value_counts())

# Ratings
print("Content Ratings:")
display(df["rating"].value_counts())

# Top 10 countries
print("Top 10 Countries:")
display(df["country"].value_counts().head(10))

# Top 10 genres
print("Top 10 Genres:")
display(df["listed_in"].value_counts().head(10))
```

```python
# Count movies and TV shows
type_counts = df["type"].value_counts()

# Create bar chart
plt.figure(figsize=(8, 5))
plt.bar(type_counts.index, type_counts.values)

# Add title and labels
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

# Show values on bars
for i, value in enumerate(type_counts.values):
    plt.text(i, value + 50, str(value), ha="center")

plt.tight_layout()
plt.show()
```

### Interpretation

The dataset contains more movies than TV shows.
This shows that movies make up the larger share
of Netflix titles in this dataset.

```python
# Count titles by release year
year_counts = df["release_year"].value_counts().sort_index()

# Create line chart
plt.figure(figsize=(10, 5))
plt.plot(year_counts.index, year_counts.values, marker='o')

# Title and labels
plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid(True)
plt.tight_layout()
plt.show()
```

### Interpretation

The graph shows the distribution of Netflix titles across different release years. It helps identify the years with higher and lower numbers of releases and reveals trends in the dataset over time.

```python
# Split countries and count each country
countries = (
    df["country"].dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
    .sort_values()
)

# Create horizontal bar chart
plt.figure(figsize=(10, 6))
plt.barh(countries.index, countries.values)

# Add title and labels
plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

### Interpretation

The graph shows the top 10 countries contributing to Netflix titles. The United States has the highest number of titles, followed by India and the United Kingdom. This indicates that these countries are highly represented in the dataset.

```python
# Count content ratings
ratings = df["rating"].value_counts().dropna()

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(ratings.index, ratings.values)

# Add title and labels
plt.title("Distribution of Netflix Content Ratings")
plt.xlabel("Content Rating")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Interpretation

The graph shows the distribution of different content ratings on Netflix. TV-MA and TV-14 are among the most frequent ratings in the dataset, while some other ratings have fewer titles. This highlights the variety of age-based content classifications available.

```python
# Filter only movies
movies = df[df["type"] == "Movie"].copy()

# Extract duration in minutes
movies["duration_mins"] = pd.to_numeric(
    movies["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)

# Remove missing duration values
movie_duration = movies["duration_mins"].dropna()

# Create histogram
plt.figure(figsize=(10, 5))
plt.hist(movie_duration, bins=25, edgecolor="black")

# Title and labels
plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

### Interpretation

The histogram shows the distribution of movie durations on Netflix. Most movies are concentrated within a certain duration range, while relatively fewer movies have very short or very long durations. This helps us understand the typical length of movies in the dataset.

```python
print("Average Duration:",
      movie_duration.mean())

print("Median Duration:",
      movie_duration.median())

print("Shortest Movie:",
      movie_duration.min())

print("Longest Movie:",
      movie_duration.max())
```
